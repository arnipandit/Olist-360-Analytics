```python
import pandas as pd
import numpy as np

customers = pd.read_csv("olist_customers_dataset.csv.zip")
orders = pd.read_csv("olist_orders_dataset.csv.zip")
order_items = pd.read_csv("olist_order_items_dataset.csv.zip")
payments = pd.read_csv("olist_order_payments_dataset.csv.zip")
products = pd.read_csv("olist_products_dataset.csv.zip")
reviews = pd.read_csv("olist_order_reviews_dataset.csv.zip")

# These are normal CSV files
sellers = pd.read_csv("olist_sellers_dataset.csv")
geolocation = pd.read_csv("olist_geolocation_dataset.csv.zip")
translation = pd.read_csv("product_category_name_translation.csv")
```

```python
def check_duplicates(df, name):

    duplicates = df.duplicated().sum()

    print("="*60)
    print(name)
    print("="*60)
    print(f"Duplicate Rows : {duplicates}")
```

```python
check_duplicates(customers,"Customers")
check_duplicates(orders,"Orders")
check_duplicates(order_items,"Order Items")
check_duplicates(payments,"Payments")
check_duplicates(reviews,"Reviews")
check_duplicates(products,"Products")
check_duplicates(sellers,"Sellers")
check_duplicates(geolocation,"Geolocation")
```

# Missing Value Analysis

```python
def missing_summary(df, name):

    missing = df.isnull().sum()
    missing = missing[missing > 0]

    print("=" * 70)
    print(name)
    print("=" * 70)

    if len(missing) == 0:
        print("No Missing Values")

    else:
        result = pd.DataFrame({
            "Missing Values": missing,
            "Percentage": round((missing / len(df)) * 100, 2)
        })

        display(result.sort_values("Percentage", ascending=False))
```

```python
missing_summary(customers, "Customers")
missing_summary(orders, "Orders")
missing_summary(order_items, "Order Items")
missing_summary(products, "Products")
missing_summary(payments, "Payments")
missing_summary(reviews, "Reviews")
missing_summary(sellers, "Sellers")
missing_summary(geolocation, "Geolocation")
```

## Missing Value Decision - Orders

The missing values in the delivery-related columns are not random.

They represent valid business scenarios such as:

- Cancelled orders
- Unavailable products
- Orders still in processing or shipment

Therefore, these missing values are retained because they provide meaningful business information.

```python
products["product_category_name"] = products["product_category_name"].fillna("Unknown")
products["product_weight_g"] = products["product_weight_g"].fillna(products["product_weight_g"].median())

products["product_length_cm"] = products["product_length_cm"].fillna(products["product_length_cm"].median())

products["product_height_cm"] = products["product_height_cm"].fillna(products["product_height_cm"].median())

products["product_width_cm"] = products["product_width_cm"].fillna(products["product_width_cm"].median())
```

```python
reviews["review_comment_title"] = reviews["review_comment_title"].fillna("No Title")
reviews["review_comment_message"] = reviews["review_comment_message"].fillna("No Comment")
```

#Convert Date Columns

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col])

reviews["review_creation_date"] = pd.to_datetime(reviews["review_creation_date"])
reviews["review_answer_timestamp"] = pd.to_datetime(reviews["review_answer_timestamp"])
```

```python
print("Customer IDs duplicated:", customers["customer_id"].duplicated().sum())

print("Order IDs duplicated:", orders["order_id"].duplicated().sum())

print("Product IDs duplicated:", products["product_id"].duplicated().sum())

print("Seller IDs duplicated:", sellers["seller_id"].duplicated().sum())
```

Conclusion: All primary keys are valid.

```python
print(orders[
        ~orders["customer_id"].isin(customers["customer_id"])
    ].shape)

print(
    order_items[
        ~order_items["order_id"].isin(orders["order_id"])
    ].shape
)

print(
    order_items[
        ~order_items["product_id"].isin(products["product_id"])
    ].shape
)
```

This means:

Every order belongs to a valid customer.
Every order item belongs to a valid order.
Every product in order_items exists in the products table.

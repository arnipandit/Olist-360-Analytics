````python
import pandas as pd
import numpy as np

customers = pd.read_csv("olist_customers_dataset.csv.zip")
orders = pd.read_csv("olist_orders_dataset.csv.zip")
order_items = pd.read_csv("olist_order_items_dataset.csv.zip")
products = pd.read_csv("olist_products_dataset.csv.zip")
payments = pd.read_csv("olist_order_payments_dataset.csv.zip")
reviews = pd.read_csv("olist_order_reviews_dataset.csv.zip")
sellers = pd.read_csv("olist_sellers_dataset.csv")
translation = pd.read_csv("product_category_name_translation.csv")
````

````python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col])
````

#Understand the Relationships

```
Customers
     │
customer_id
     │
     ▼
Orders
     │
order_id
     │
     ▼
Order Items
     │
product_id
     ▼
Products

````python
customer360 = orders.copy()
customer360.head()
````

````python
customer360 = customer360.merge(
    customers,
    on="customer_id",
    how="left"
)
customer360.head()
````

````python
customer360 = customer360.merge(
    order_items,
    on="order_id",
    how="left"
)
customer360.shape
````

````python
customer360 = customer360.merge(
    products,
    on="product_id",
    how="left"
)
````

````python
customer360 = customer360.merge(
    payments,
    on="order_id",
    how="left"
)
````

````python
customer360 = customer360.merge(
    reviews,
    on="order_id",
    how="left"
)
````

````python
customer360.head()
````

````python
customer360.info()
````

````python
customer360.shape
````

````python
customer360.to_csv("customer360_master.csv", index=False)
````

Date Features

````python
customer360['Purchase_Year'] = customer360['order_purchase_timestamp'].dt.year
customer360['Purchase_Month'] = customer360['order_purchase_timestamp'].dt.month_name()
customer360['Purchase_Quarter'] = customer360['order_purchase_timestamp'].dt.quarter
customer360['Purchase_Day'] = customer360['order_purchase_timestamp'].dt.day_name()
````

Delivery Features

````python
customer360['Delivery_Days'] = (
    customer360['order_delivered_customer_date']
    - customer360['order_purchase_timestamp']
).dt.days

customer360['Delivery_Delay'] = (
    customer360['order_delivered_customer_date']
    - customer360['order_estimated_delivery_date']
).dt.days

customer360['Delivery_Status'] = np.where(
    customer360['Delivery_Delay'] > 0,
    'Late',
    'On Time/Early'
)
````

Customer Features

````python
customer360.columns.tolist()
````

````python
(customer360['customer_unique_id_x'] == customer360['customer_unique_id_y']).all()
````

````python
customer360.drop(
    columns=[
        'customer_unique_id_y',
        'customer_zip_code_prefix_y',
        'customer_city_y',
        'customer_state_y'
    ],
    inplace=True
)
````

````python
customer360['Customer_Total_Orders'] = customer360.groupby(
    'customer_unique_id'
)['order_id'].transform('nunique')

customer360['Customer_Total_Spend'] = customer360.groupby(
    'customer_unique_id'
)['payment_value'].transform('sum')

customer360['Repeat_Customer'] = np.where(
    customer360['Customer_Total_Orders'] > 1,
    'Yes',
    'No'
)
````

Review Features

````python
customer360['Review_Category'] = np.where(
    customer360['review_score'] >= 4,
    'Positive',
    'Negative'
)
````

````python
customer360.to_csv('customer360_final.csv', index=False)
````
